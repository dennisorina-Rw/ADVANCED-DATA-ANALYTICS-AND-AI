## **Spearman Rank-Order Correlation**

### **Definition**

Spearman Rank-Order Correlation, commonly denoted by **Spearman's rho (ρ or (r_s))**, is a **non-parametric statistical method** used to measure the **strength and direction of a monotonic relationship** between two variables. 

Unlike Pearson correlation, Spearman correlation does not require the variables to be normally distributed or linearly related. Instead, it uses the **ranks** of the observations rather than their actual values.

Spearman correlation is particularly useful when the data are **ordinal**, contain **outliers**, or violate the assumptions required for Pearson correlation. The coefficient ranges from **−1 to +1**, where the sign indicates the direction of the relationship and the magnitude indicates its strength.

### **When to Use Spearman Rank Correlation**

Spearman correlation is appropriate when:

* The variables are measured on an **ordinal**, **interval**, or **ratio** scale.
* The relationship between variables is **monotonic** but not necessarily linear.
* The data are **not** normally distributed.
* The dataset contains influential outliers that may distort Pearson correlation.
* The observations can be ranked meaningfully.
  
### **Assumptions of Spearman Rank Correlation**

Although Spearman correlation is less restrictive than Pearson correlation, several assumptions should still be satisfied.

#### **1. Paired Observations**

Each observation in one variable must have a corresponding observation in the other variable. Every pair should come from the same individual or experimental unit.

#### **2. Independent Observations**

The observations should be independent of one another. The measurement of one participant or subject should not influence another.

#### **3. Variables Can Be Ranked**

The variables should be at least ordinal in nature so that observations can be arranged from the smallest to the largest value.

#### **4. Monotonic Relationship**

The relationship between the variables should be monotonic. This means that as one variable increases, the other variable consistently tends to increase or decrease, although the rate of change does not have to remain constant.

#### **5. No Requirement for Normality**

Unlike Pearson correlation, Spearman correlation does not require either variable to follow a normal distribution because it operates on ranked data.

### **Mathematical Expression**

When there are **no tied ranks**, Spearman's rank correlation coefficient is calculated as:

$$
r_s = 1 - \frac{6\sum d_i^2}{n(n^2-1)}
$$

Where:

* $r_s$ = Spearman rank correlation coefficient
* $d_i$ = Difference between the ranks of the two variables for the (i^{th}) observation
* $n$ = Number of paired observations
* $\sum d_i^2$ = Sum of the squared differences between ranks

### **Interpretation of Spearman Correlation Coefficient**

The value of Spearman's rho ranges from **−1 to +1**.

### **Hypothesis Testing**

The hypotheses for Spearman correlation are:

* **Null hypothesis $H_0$**: There is no monotonic association between the two variables $\rho = 0$.
* **Alternative hypothesis $H_1$**: There is a significant monotonic association between the two variables $\rho \neq 0$.

The decision is based on the **p-value**:

* If **p ≤ 0.05**, reject $H_0$; there is a statistically significant monotonic relationship.
* If **p > 0.05**, fail to reject $H_0$; there is insufficient evidence of a monotonic relationship.

### **Advantages**

Spearman Rank Correlation has several advantages:

* It does not require normally distributed data.
* It is suitable for ordinal data.
* It is less sensitive to outliers than Pearson correlation.
* It can detect monotonic relationships even when they are not linear.
* It is simple to compute and interpret.

### **Limitations**

Despite its flexibility, Spearman correlation has some limitations:

* It measures only **monotonic** relationships and may fail to detect more complex associations.
* Ranking data may result in some loss of information compared with using the original measurements.
* When many tied ranks are present, adjustments are required, and the simple formula is no longer exact.
  
### **Summary**

Spearman Rank-Order Correlation is a robust non-parametric measure of association that evaluates the strength and direction of a monotonic relationship between two variables using ranked data. It is widely used when data are ordinal, non-normal, or contain outliers, making it an excellent alternative to Pearson correlation in many practical research situations.

## **Covariance**

### **Definition**

Covariance is a statistical measure that indicates the **direction of the linear relationship** between two variables. It shows whether two variables tend to increase or decrease together.

- **Positive covariance:** As one variable increases, the other tends to increase.
- **Negative covariance:** As one variable increases, the other tends to decrease.
- **Zero covariance:** There is no consistent linear relationship between the variables.

Unlike correlation, the magnitude of covariance depends on the units of measurement, making it difficult to compare across different datasets.

## **Formula for Population Covariance**

$$
\operatorname{Cov}(X,Y)=\frac{\sum_{i=1}^{N}(X_i-\mu_X)(Y_i-\mu_Y)}{N}
$$

where:

- $X_i, Y_i$ = individual observations
- $\mu_X, \mu_Y$ = population means
- $N$ = population size

## **Formula for Sample Covariance**

$$
s_{XY}=\frac{\sum_{i=1}^{n}(X_i-\bar{X})(Y_i-\bar{Y})}{n-1}
$$

where:

- $\bar{X}, \bar{Y}$ = sample means
- $n$ = sample size

## **Interpretation**

| Covariance Value | Interpretation |
|------------------|----------------|
| $\operatorname{Cov}(X,Y) > 0$ | Positive relationship |
| $\operatorname{Cov}(X,Y) < 0$ | Negative relationship |
| $\operatorname{Cov}(X,Y) = 0$ | No linear relationship |

## **Example**

Consider the following data:

| Hours Studied | Exam Score |
|---------------|-----------:|
| 2 | 50 |
| 4 | 60 |
| 6 | 70 |
| 8 | 80 |
| 10 | 90 |

As study hours increase, exam scores also increase.

- Covariance $>0$
- Indicates a positive relationship between study time and exam performance.

## **Properties of Covariance**

### **1. Symmetry**

$$
\operatorname{Cov}(X,Y)=\operatorname{Cov}(Y,X)
$$

The covariance between two variables is the same regardless of their order.

### **2. Covariance of a Variable with Itself**

$$
\operatorname{Cov}(X,X)=\operatorname{Var}(X)
$$

The covariance of a variable with itself is equal to its variance.

### **3. Independent Variables Have Zero Covariance**

If $X$ and $Y$ are independent, then

$$
\operatorname{Cov}(X,Y)=0
$$

> **Note:** Zero covariance does **not** necessarily imply independence. Variables may have a nonlinear relationship while still having zero covariance.

### **4. Effect of Scaling**

If

$$
U=aX,\qquad V=bY
$$

then

$$
\operatorname{Cov}(U,V)=ab\operatorname{Cov}(X,Y)
$$

This shows that covariance changes when variables are multiplied by constants.

## **Advantages**

- Measures the direction of a linear relationship.
- Forms the basis for correlation and regression analysis.
- Used extensively in finance to study how assets move together.
- Helps understand joint variability between variables.
- Essential in multivariate statistical techniques.

## **Limitations**

- Depends on the units of measurement.
- The magnitude has no standardized interpretation.
- Difficult to compare across datasets with different scales.
- Does not indicate the strength of the relationship.
- Does not imply causation.

## **Covariance vs. Correlation**

| Feature | Covariance | Correlation |
|---------|------------|-------------|
| Range | Unbounded | $-1$ to $1$ |
| Unit | Depends on units | Unit-free |
| Indicates direction | Yes | Yes |
| Indicates strength | No | Yes |
| Easy to interpret | No | Yes |
| Commonly used | Less frequently | Very frequently |

## **Applications**

- **Finance:** Portfolio optimization and risk management.
- **Machine Learning:** Covariance matrices in Principal Component Analysis (PCA).
- **Statistics:** Regression, multivariate analysis, and hypothesis testing.
- **Economics:** Studying relationships among economic indicators.
- **Health Sciences:** Investigating relationships among clinical variables.

## **Covariance Matrix**

For multiple variables, covariance is summarized in a covariance matrix.

For variables $X_1, X_2, \ldots, X_p$,

$$
\Sigma=
\begin{bmatrix}
\operatorname{Var}(X_1) & \operatorname{Cov}(X_1,X_2) & \cdots \\
\operatorname{Cov}(X_2,X_1) & \operatorname{Var}(X_2) & \cdots \\
\vdots & \vdots & \ddots
\end{bmatrix}
$$

where:

- The **diagonal elements** represent the variances.
- The **off-diagonal elements** represent the covariances between pairs of variables.

## **Key Takeaways**

- Covariance measures the **direction** of the linear relationship between two variables.
- A **positive covariance** indicates that variables tend to increase or decrease together.
- A **negative covariance** indicates that one variable tends to increase while the other decreases.
- A covariance of **zero** indicates no linear relationship.
- Because covariance depends on the units of measurement, it is difficult to compare across different datasets.
- Correlation is a standardized form of covariance and is generally preferred for measuring the **strength** of a relationship.
- Covariance is a fundamental concept in regression analysis, principal component analysis (PCA), portfolio optimization, and many other multivariate statistical methods.

### **Simple Linear Regression Model**
## **Introduction**

Simple Linear Regression (SLR) is one of the most widely used statistical techniques for examining and modeling the relationship between **one independent (predictor) variable** and **one dependent (response) variable**. The method estimates how changes in the predictor variable influence the response variable and provides a mathematical equation that can be used for prediction.

The objective of simple linear regression is to determine the **best-fitting straight line** through a set of observed data points by minimizing the sum of the squared differences between the observed values and the predicted values. This approach is known as the **Ordinary Least Squares (OLS)** method.

### **History of Simple Linear Regression**

The foundations of linear regression date back to the early nineteenth century.

* **Adrien-Marie Legendre (1805)** introduced the **Least Squares Method**, which estimates model parameters by minimizing the sum of squared residuals.
* **Carl Friedrich Gauss (1809)** further developed the theoretical properties of least squares estimation and demonstrated its statistical efficiency under certain assumptions.
* The Gauss–Markov Theorem later established that OLS estimators are the **Best Linear Unbiased Estimators (BLUE)** when the classical regression assumptions are satisfied.

Today, simple linear regression is extensively applied in economics, medicine, agriculture, engineering, finance, business analytics, environmental science, and social sciences.

### **Purpose of Simple Linear Regression**

Simple linear regression is used to:

* Examine whether a linear relationship exists between two continuous variables.
* Quantify the direction and magnitude of the relationship.
* Predict the value of a dependent variable from an independent variable.
* Assess whether the predictor significantly explains variation in the response variable.

##### **Mathematical Model**

The population simple linear regression model is expressed as

$$
Y_i=\beta_0+\beta_1X_i+\varepsilon_i,\qquad i=1,2,\ldots,n
$$

where

* $Y_i$ = dependent (response) variable
* $X_i$ = independent (predictor) variable
* $\beta_0$ = population intercept
* $\beta_1$ = population slope (regression coefficient)
* $\varepsilon_i$ = random error term
* $n$ = number of observations

The fitted sample regression equation is

$$
\hat{Y}=b_0+b_1X
$$

where

* $\hat{Y}$ = predicted value of the dependent variable
* $b_0$ = estimated intercept
* $b_1$ = estimated slope

### **Interpretation of the Coefficients**

#### **Intercept $\beta_0$**

The intercept represents the expected value of the dependent variable when the independent variable is equal to zero.

Although it has practical meaning in some studies, in others it simply provides the mathematical starting point of the regression line.

### Slope $\beta_1$

The slope measures the average change in the dependent variable associated with a one-unit increase in the independent variable.

* If $b_1>0$, the relationship is positive.
* If $b_1<0$, the relationship is negative.
* If $b_1=0$, there is no linear relationship.

# Estimation of Regression Coefficients

The regression coefficients are estimated using the **Ordinary Least Squares (OLS)** method, which minimizes the sum of squared residuals.

## Estimated Slope

$$
b_1=
\frac{\sum_{i=1}^{n}(X_i-\bar X)(Y_i-\bar Y)}
{\sum_{i=1}^{n}(X_i-\bar X)^2}
$$

where

* $b_1$ = estimated slope
* $\bar X$ = mean of the predictor variable
* $\bar Y$ = mean of the response variable

## Estimated Intercept

$$
b_0=\bar Y-b_1\bar X
$$

where

* $b_0$ = estimated intercept
* $\bar X$ = sample mean of (X)
* $\bar Y$ = sample mean of (Y)

## Residual

A residual is the difference between the observed and predicted values.

$$
e_i=Y_i-\hat Y_i
$$

Residuals measure prediction errors and are fundamental in assessing model adequacy.

# Assumptions of Simple Linear Regression

The validity of the regression model depends on several assumptions.

### 1. Linearity

There should be a linear relationship between the independent and dependent variables.

### 2. Independence of Observations

Observations should be independent, meaning that one observation does not influence another.

### 3. Homoscedasticity

The variance of the residuals should remain constant across all levels of the predictor variable.

### 4. Normality of Residuals

Residuals should be approximately normally distributed, particularly for hypothesis testing and confidence intervals.

### 5. No Influential Outliers

Extreme observations should not unduly influence the fitted regression model.

### 6. Measurement Reliability

Both variables should be measured accurately, with minimal measurement error in the predictor variable.

---

# Hypothesis Testing

The significance of the regression slope is tested using

* **Null hypothesis**

$$
H_0:\beta_1=0
$$

There is no linear relationship between $X$ and $Y$.

* **Alternative hypothesis**

$$
H_1:\beta_1\neq0
$$

There is a significant linear relationship between (X) and (Y).

The decision is based on the **t-test** and the associated **p-value**.

* If **p ≤ 0.05**, reject (H_0).
* If **p > 0.05**, fail to reject (H_0).

# Coefficient of Determination $R^2$

The coefficient of determination measures the proportion of the variation in the dependent variable explained by the regression model.

$$
R^2=
\frac{\text{Regression Sum of Squares (SSR)}}
{\text{Total Sum of Squares (SST)}}
=1-\frac{\text{SSE}}{\text{SST}}
$$

where

* **SST** = Total Sum of Squares
* **SSR** = Regression Sum of Squares
* **SSE** = Error (Residual) Sum of Squares

The value of (R^2) ranges from **0 to 1**.

* $R^2=0$: the predictor explains none of the variation.
* $R^2=1$: the predictor perfectly explains the variation.

# Advantages

Simple linear regression offers several advantages:

* It is easy to understand and interpret.
* It quantifies the relationship between two variables.
* It enables prediction of future observations.
* It forms the foundation of multiple regression and many advanced statistical models.
* It provides inferential statistics such as confidence intervals and hypothesis tests.

# Limitations

Despite its usefulness, simple linear regression has some limitations.

* It models only one predictor variable.
* It assumes a linear relationship between variables.
* It is sensitive to influential outliers.
* Violations of regression assumptions may produce biased or inefficient estimates.
* Correlation or regression does not imply causation.

# Applications

Simple linear regression is commonly used in:

* Predicting sales from advertising expenditure.
* Estimating crop yield from rainfall.
* Examining the relationship between age and blood pressure.
* Modeling house prices based on size.
* Predicting employee salaries from years of experience.
* Assessing relationships in healthcare, economics, agriculture, finance, engineering, and environmental science.

Simple Linear Regression is a statistical technique that models the relationship between one predictor and one response variable using a straight-line equation estimated by the Ordinary Least Squares method. It provides estimates of the intercept and slope, allows prediction, measures the strength of the relationship through (R^2), and supports hypothesis testing. When its assumptions are satisfied, the OLS estimators are unbiased, efficient, and form the basis for many advanced statistical and machine learning models.

### **define labels to use for output of White's test**

```python
labels = ['Test Statistic', 'Test Statistic p-value', 'F-Statistic', 'F-Test p-value']

#print results of White's test
dict(zip(labels, white_test))
```

### **Here is how to interpret the output:**

The test statistic is t-statistic = 17.937
The corresponding p-value is 0.0359.
White’s test uses the following null and alternative hypotheses:

* Null (H0): Homoscedasticity is present (residuals are equally scattered and have constant variance)
* Alternative (HA): Heteroscedasticity is present (residuals are not equally scattered and have none constant variance)

Since the p-value is less than 0.05 we reject the null hypothesis.
